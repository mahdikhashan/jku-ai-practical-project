```python
import torch
```

```python
cfg = create_lizard_attention_config(
    batch_size=2,
    seq_len=8192 * 4,
    hidden_size=256,
    num_heads=4,
    dtype="float32",
    device="cuda" if torch.cuda.is_available() else "cpu",
)
```

```python
from modules.lizard_attention_block_triton import LizardAttentionBlock
```

```python
torch_dtype = cfg.dtype.to_torch()
head_dim = cfg.hidden_size // cfg.num_heads

print(f"Configuring model with: {cfg}")
print(f"Head Dimension: {head_dim}")

model = (
    LizardAttentionBlock(
        d_model=cfg.hidden_size, n_heads=cfg.num_heads, window_size=32, alpha=0.5, m=4
    )
    .to(cfg.device)
    .to(torch_dtype)
)

q = torch.randn(
    cfg.batch_size,
    cfg.num_heads,
    cfg.seq_len,
    head_dim,
    device=cfg.device,
    dtype=torch_dtype,
)
k = torch.randn(
    cfg.batch_size,
    cfg.num_heads,
    cfg.seq_len,
    head_dim,
    device=cfg.device,
    dtype=torch_dtype,
)
v = torch.randn(
    cfg.batch_size,
    cfg.num_heads,
    cfg.seq_len,
    head_dim,
    device=cfg.device,
    dtype=torch_dtype,
)

output = model(q, k, v)

print(f"Input shape:  {q.shape}")
print(f"Output shape: {output.shape}")

assert q.shape == output.shape
```

```python
def benchmark():
    model(q, k, v)

benchmark()
```

```python
%timeit -r 5 -n 100 benchmark()
```

```python
import torch
import triton
from modules.lizard_attention_block_triton import LizardAttentionBlock

device = "cuda"
L = 32768  # 32k
D = 64     # Head Dim
H = 4      # Num Heads

model = LizardAttentionBlock(d_model=D*H, n_heads=H).to(device)
q = torch.randn(1, H, L, D, device=device, dtype=torch.float16)
k = torch.randn(1, H, L, D, device=device, dtype=torch.float16)
v = torch.randn(1, H, L, D, device=device, dtype=torch.float16)

print(f"Benchmarking L={L}...")

for _ in range(10):
    _ = model(q, k, v)
torch.cuda.synchronize()

starter, ender = torch.cuda.Event(enable_timing=True), torch.cuda.Event(enable_timing=True)
timings = []

for _ in range(100):
    starter.record()
    _ = model(q, k, v)
    ender.record()
    torch.cuda.synchronize()
    timings.append(starter.elapsed_time(ender))

import numpy as np
print(f"Latency: {np.mean(timings):.3f} ms ± {np.std(timings):.3f}")
print(f"Throughput: {1000/np.mean(timings):.1f} batches/sec")
```
